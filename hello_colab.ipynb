# Hello Colab

This notebook is a minimal example for Google Colab. It defines small `add(a, b)` and `multiply(a, b)` functions, prints a greeting, and prints the results of adding and multiplying two numbers. It uses only core Python, so there is nothing to install.

Run the cell below with **Shift+Enter**, or choose **Runtime > Run all**.

```python
def add(a, b):
    """Return the sum of a and b."""
    return a + b


def multiply(a, b):
    """Return the product of a and b."""
    return a * b


print("Hello from Colab!")
print(add(2, 3))
print(multiply(2, 3))
```
